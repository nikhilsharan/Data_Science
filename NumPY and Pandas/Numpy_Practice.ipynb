```python
import numpy as np


# Generate seller data: 10 rows, 3 columns
data = np.random.randint([0, 0, 0],[100001, 1001, 101],size=(10, 3))

print(data)
# Seller IDs
seller_ids = np.arange(1, 11).reshape(10, 1)

# # Combine ID with data
dataset = np.hstack((seller_ids, data))

# print("📦 Seller Dataset (ID | Sales | Orders | Rating):")
print(dataset)
```

```python
print(dataset[4,::])
```

```python
print(dataset[2:7,::])
```

```python
print(dataset[::,0:2])
```

```python
print(dataset[::,0:3:2])
```

```python
print(dataset)
```

```python
print(dataset[dataset[:,3]>=20])
```

```python
print(dataset[dataset[:,3]>=20,1])
```

```python
print(np.mean(dataset[dataset[:,3]>=20,1]))
```

```python
seller_id = int(input("Enter seller id"))

print(dataset[dataset[:,0]==seller_id])
```
