# 🔍 Linear Search

## 📘 What is Linear Search?
Linear Search is the simplest searching technique where we check **each element one by one** until we find the target element or reach the end of the list.

---

## 🎯 When is Linear Search Used?
Use Linear Search when:
- The list is **unsorted**
- The list is **small in size**
- You need a simple searching method
- No extra memory should be used (O(1) space)

---

## 🧠 How Linear Search Works (Step-by-Step)
1. Start from the **first element** of the list.
2. Compare it with the **target value**.
3. If it matches → return the **index**.
4. If it does not match → move to the **next element**.
5. Continue until:
   - The target is found, or  
   - The end of the list is reached.
6. If not found → return **-1**.

---

## ⏱️ Time & Space Complexity

| Case          | Time Complexity |
|---------------|------------------|
| **Best Case** | O(1)             |
| **Worst Case**| O(n)             |
| **Average**   | O(n)             |
| **Space**     | O(1)             |

---

## 👍 Advantages
- Works on **sorted and unsorted** lists  
- Very easy to understand  
- No extra memory needed  

## 👎 Disadvantages
- Slow for **large** lists  
- Inefficient compared to Binary Search for sorted data  

---


```python
def linear_search(arr, target):
    for i, value in enumerate(arr):
        if value == target:
            return i
    return -1

# Example
arr = [10, 4, 2, 7, 9]
print("Index of 7:", linear_search(arr, 7))   # Expected: 3
print("Index of 5:", linear_search(arr, 5))   # Expected: -1
```

# 🔎 Binary Search

## 📘 What is Binary Search?
Binary Search is an efficient algorithm used to find an element in a **sorted list** by repeatedly dividing the search interval in half.

It eliminates half of the remaining elements each step, making it much faster than Linear Search.

---

## ⚠️ Important Requirement
Binary Search works **only on sorted arrays**  
(ascending or descending, as long as you know the order).

---

## ⚡ Why is Binary Search Faster?
Instead of checking elements one by one, it jumps directly to the **middle element** and decides which half to continue searching in.

So instead of O(n), it works in **O(log n)** time.

---

## 🧠 How Binary Search Works (Step-by-Step)

1. Set two pointers:
   - `low` → start of array  
   - `high` → end of array  

2. Find the **middle index**  
   `mid = (low + high) // 2`

3. Compare the **mid element** with the target:
   - If `arr[mid] == target` → return the index  
   - If `arr[mid] < target` → search the **right half**  
   - If `arr[mid] > target` → search the **left half**  

4. Repeat the process until:
   - Target is found, or  
   - `low` becomes greater than `high` (target not found)

---

## ⏱️ Time & Space Complexity

| Case | Time Complexity |
|------|------------------|
| Best Case | O(1) |
| Worst Case | O(log n) |
| Average Case | O(log n) |
| Space (Iterative) | O(1) |
| Space (Recursive) | O(log n) |

---

## 👍 Advantages
- Extremely fast for large sorted lists  
- Reduces time drastically compared to linear search  
- Very efficient: O(log n) time  

## 👎 Disadvantages
- Works **only** on sorted lists  
- More complex to implement  
- Recursive version uses stack space  

---

## ▶️ Python Code (Iterative + Recursive)

```python
# -------------------------------
# Iterative Binary Search
# -------------------------------

def binary_search_iterative(arr, target):
    """
    Perform binary search using an iterative approach.
    Array must be sorted.
    Returns index if found, else -1.
    """
    low, high = 0, len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1


# -------------------------------
# Recursive Binary Search
# -------------------------------

def binary_search_recursive(arr, target, low, high):
    """
    Recursive implementation of binary search.
    Returns index of target, or -1 if not found.
    """
    if low > high:
        return -1

    mid = (low + high) // 2

    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search_recursive(arr, target, mid + 1, high)
    else:
        return binary_search_recursive(arr, target, low, mid - 1)


# -------------------------------
# Example Usage
# -------------------------------
arr = [10, 20, 30, 40, 50]
target = 40

print("Iterative:", binary_search_iterative(arr, target))        # Expected: 3
print("Recursive:", binary_search_recursive(arr, target, 0, len(arr) - 1))  # Expected: 3
```

# 🔄 Selection Sort

## 📘 What is Selection Sort?
Selection Sort is a simple comparison-based sorting algorithm that repeatedly selects the **smallest element** (for ascending order) from the unsorted portion of the list and places it in its correct position.

It divides the list into two parts:
1. **Sorted part** (left side)
2. **Unsorted part** (right side)

---

## 🧠 How Selection Sort Works (Step-by-Step)

1. Start from index `0`.
2. Assume the first unsorted element is the **minimum**.
3. Compare it with every other element in the unsorted part.
4. Find the **exact minimum element**.
5. Swap the minimum element with the first unsorted index.
6. Move the boundary of the sorted region one step to the right.
7. Repeat until the list is fully sorted.

---

## 🔍 Example (Ascending Order)

Unsorted list:  
`[64, 25, 12, 22, 11]`

Pass-by-pass:
- Pass 1 → smallest is 11 → swap with index 0  
  Result: `[11, 25, 12, 22, 64]`
- Pass 2 → smallest is 12 → swap with index 1  
  Result: `[11, 12, 25, 22, 64]`
- Pass 3 → smallest is 22 → swap with index 2  
  Result: `[11, 12, 22, 25, 64]`
- Pass 4 → smallest is 25 → already at position  
  Result: `[11, 12, 22, 25, 64]`

---

## ⏱️ Time & Space Complexity

| Case          | Time Complexity |
|---------------|------------------|
| Best Case     | O(n²) |
| Worst Case    | O(n²) |
| Average Case  | O(n²) |
| Space         | O(1) (in-place sorting) |

---

## 👍 Advantages
- Easy to understand  
- Very few swaps compared to bubble sort  
- In-place algorithm → no extra memory required  

## 👎 Disadvantages
- Slow for large lists  
- Always performs O(n²) comparisons  
- Not stable (relative order not always preserved)  

---

## ▶️ Python Code

```python
# Selection Sort Implementation

def selection_sort(arr):
    """
    Sorts the array in ascending order using Selection Sort.
    Works by repeatedly finding the minimum element
    from the unsorted part and placing it at the beginning.
    """
    n = len(arr)

    for i in range(n):
        # Assume the current index has the minimum value
        min_index = i

        # Find the actual minimum in the remaining unsorted part
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        # Swap the found minimum with the first unsorted element
        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr


# Example usage
arr = [64, 25, 12, 22, 11]
print("Sorted array:", selection_sort(arr))
```

# 🧩 Insertion Sort

## 📘 What is Insertion Sort?
Insertion Sort is a simple and intuitive sorting algorithm that works the same way you sort playing cards in your hand.  
You build the final sorted array **one element at a time**.

At each step:
- Remove one element from the input
- Compare it with elements in the sorted part (left side)
- Insert it at the correct position

---

## 🧠 How Insertion Sort Works (Step-by-Step)

1. Start from index `1` (index `0` is already considered sorted).
2. Take the current element → call it **key**.
3. Compare `key` with elements to its left.
4. Shift larger elements **one position to the right**.
5. Insert `key` in its correct position.
6. Move to the next index and repeat.

---

## 🔍 Example

For the list:  
`[9, 5, 1, 4, 3]`

Step-by-step:
- Insert 5 → `[5, 9, 1, 4, 3]`
- Insert 1 → `[1, 5, 9, 4, 3]`
- Insert 4 → `[1, 4, 5, 9, 3]`
- Insert 3 → `[1, 3, 4, 5, 9]`

---

## ⏱️ Time & Space Complexity

| Case | Time Complexity |
|------|------------------|
| Best Case | O(n) (already sorted) |
| Worst Case | O(n²) |
| Average Case | O(n²) |
| Space | O(1) (in-place) |

---

## 👍 Advantages
- Great for **small datasets**
- Efficient for **almost sorted** lists
- Simple implementation
- Stable sorting (preserves order of equal elements)

## 👎 Disadvantages
- Slow for large datasets (O(n²))
- Lots of shifts for reversed arrays

---

## ▶️ Python Code

```python
# Insertion Sort Implementation

def insertion_sort(arr):
    """
    Sorts the array using the Insertion Sort algorithm.
    Builds the sorted list one element at a time by
    inserting each element into its correct position.
    """
    for i in range(1, len(arr)):
        key = arr[i]          # element to be inserted
        j = i - 1

        # Shift elements that are greater than key to the right
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        # Insert key into correct position
        arr[j + 1] = key

    return arr


# Example usage
arr = [9, 5, 1, 4, 3]
print("Sorted array:", insertion_sort(arr))
```

# 🧵 Merge Sort

## 📘 What is Merge Sort?
Merge Sort is a **divide-and-conquer** sorting algorithm.  
It works by splitting the array into halves, sorting each half recursively, and then merging the sorted halves.

It is one of the most efficient general-purpose sorting algorithms with **O(n log n)** time complexity.

---

## 🧠 How Merge Sort Works (Step-by-Step)

### 1. **Divide**
- Split the list into two halves:
  - Left half
  - Right half

### 2. **Conquer**
- Recursively apply Merge Sort on each half
- Base Case: If the list has **0 or 1 element**, it is already sorted

### 3. **Combine**
- Merge two sorted halves into a single sorted list
- Compare smallest elements of each half and build the final sorted array

---

## 🔍 Example

List:  
`[38, 27, 43, 3, 9, 82, 10]`

Process:
- Split → `[38, 27, 43]` and `[3, 9, 82, 10]`
- Recursively split until single elements
- Merge step-by-step:
  - `[38] + [27] → [27, 38]`
  - `[27, 38] + [43] → [27, 38, 43]`
  - Continue merging all halves…
- Final result:  
  `[3, 9, 10, 27, 38, 43, 82]`

---

## ⏱️ Time & Space Complexity

| Case | Time Complexity |
|------|------------------|
| Best Case | O(n log n) |
| Worst Case | O(n log n) |
| Average Case | O(n log n) |
| Space | O(n) (needs extra temporary arrays) |

---

## 👍 Advantages
- Very efficient (O(n log n))  
- Stable sorting algorithm  
- Works great for large datasets  

## 👎 Disadvantages
- Requires extra memory → O(n)  
- Slower than quicksort in practice (sometimes)  

---

## ▶️ Python Code

```python
def merge_sort(arr):
    # Base case: A list of 0 or 1 elements is already sorted
    if len(arr) <= 1:
        return arr
    
    # Divide the array into two halves
    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])
    
    # Merge the sorted halves
    return merge(left_half, right_half)

def merge(left, right):
    sorted_list = []
    i = j = 0

    # Compare and merge
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            sorted_list.append(left[i])
            i += 1
        else:
            sorted_list.append(right[j])
            j += 1
    
    # Add any remaining elements
    sorted_list.extend(left[i:])
    sorted_list.extend(right[j:])
    
    return sorted_list


# Example usage
arr = [38, 27, 43, 3, 9, 82, 10]
print("Original:", arr)
print("Sorted:", merge_sort(arr))
```

# ⚡ Quick Sort

## 📘 What is Quick Sort?
Quick Sort is an efficient, comparison-based, divide-and-conquer sorting algorithm.  
It works by selecting a **pivot element** and partitioning the array into:
- Elements **less than** the pivot  
- Elements **greater than** the pivot  

Then, it recursively sorts the two partitions.

---

## 🧠 How Quick Sort Works (Step-by-Step)

1. **Choose a pivot**  
   Common strategies:  
   - First element  
   - Last element  
   - Middle element  
   - Random element  

2. **Partition the array**  
   Rearrange elements so that:
   - Values `< pivot` go to the left  
   - Values `> pivot` go to the right  

3. **Recursively apply Quick Sort**  
   Apply the same process to the left and right halves.

4. **Combine results**  
   Since sorting is done in-place, no merging step is required.

---

## ⏱️ Time & Space Complexity

| Case | Time Complexity |
|------|-----------------|
| Best Case | O(n log n) |
| Average Case | O(n log n) |
| Worst Case | O(n²) (pivot always smallest or largest) |
| Space Complexity | O(log n) recursion stack |

---

## 👍 Advantages
- Excellent average performance  
- In-place sorting (no extra array required)  
- Very fast in practice  

## 👎 Disadvantages
- Worst-case O(n²) if pivot poorly chosen  
- Recursive approach uses stack space  

---

## 🔍 Example

Array: `[10, 7, 8, 9, 1, 5]`  

Steps:
1. Choose pivot = **5**  
2. Partition → `[1] [5] [10, 7, 8, 9]`  
3. Recursively sort left and right  
4. Final sorted array → `[1, 5, 7, 8, 9, 10]`

---

## ▶️ Python Code

```python
# Quick Sort Implementation

def quick_sort(arr):
    """
    Sorts an array using the Quick Sort algorithm.
    Returns a new sorted array.
    """
    if len(arr) <= 1:
        return arr  # Base case: already sorted
    
    # Choose pivot (middle element)
    pivot = arr[len(arr) // 2]
    
    # Partition into three lists
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    # Recursively sort left and right, then combine
    return quick_sort(left) + middle + quick_sort(right)


# Example usage
arr = [10, 7, 8, 9, 1, 5]
print("Original array:", arr)
print("Sorted array:", quick_sort(arr))
```

# 🔺 Heapify & Heap Sort

## 📘 What is Heap Sort?
Heap Sort is a **comparison-based sorting algorithm** that uses a **binary heap data structure**.  
It works in two main steps:
1. Build a **max heap** (for ascending sort) from the array.
2. Repeatedly extract the maximum element from the heap and place it at the end of the array.

---

## 🧠 How Heap Sort Works (Step-by-Step)

### Step 1: Build Max Heap
- A max heap is a binary tree where the parent node is **greater than its children**.
- Start from the last non-leaf node and heapify each node.

### Step 2: Extract Elements
- Swap the first element (max) with the last element.
- Reduce heap size by one.
- Heapify the root again.
- Repeat until the heap size becomes 1.

---

## 🔍 Example

Array: `[12, 11, 13, 5, 6, 7]`

1. Build max heap → `[13, 12, 7, 5, 6, 11]`  
2. Swap max with last element → `[11, 12, 7, 5, 6, 13]`  
3. Heapify root → `[12, 11, 7, 5, 6, 13]`  
4. Repeat until fully sorted → `[5, 6, 7, 11, 12, 13]`

---

## ⏱️ Time & Space Complexity

| Case | Time Complexity |
|------|-----------------|
| Best Case | O(n log n) |
| Worst Case | O(n log n) |
| Average Case | O(n log n) |
| Space Complexity | O(1) (in-place) |

---

## 👍 Advantages
- In-place sorting (no extra array)  
- Guaranteed O(n log n) time complexity  
- No worst-case slowdowns like quick sort

## 👎 Disadvantages
- Not stable (equal elements may change order)  
- Slightly more complex than simple sorts (insertion/selection)

---

## ▶️ Python Code

```python
# Heapify function
def heapify(arr, n, i):
    """
    Maintains the max-heap property for a subtree rooted at index i.
    n: size of heap
    """
    largest = i        # Initialize largest as root
    left = 2 * i + 1   # Left child index
    right = 2 * i + 2  # Right child index

    # Check if left child exists and is greater than root
    if left < n and arr[left] > arr[largest]:
        largest = left

    # Check if right child exists and is greater than current largest
    if right < n and arr[right] > arr[largest]:
        largest = right

    # Swap and continue heapifying if root is not largest
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)


# Heap Sort function
def heap_sort(arr):
    n = len(arr)

    # Build max heap
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # Extract elements one by one
    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]  # Swap
        heapify(arr, i, 0)               # Heapify root

    return arr


# Example usage
arr = [12, 11, 13, 5, 6, 7]
print("Original array:", arr)
print("Sorted array:", heap_sort(arr))
```

# 🏷️ Sorting Custom Objects with a Key

## 📘 Scenario
Suppose we have a class `Student` with the following attributes:
- `name` → string
- `age` → integer
- `grade` → float

We want to sort a list of students by different criteria: name, age, or grade.

---


```python

## 🔍 Example: Define the Class and List

class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def __repr__(self):
        return f"{self.name} (Age: {self.age}, Grade: {self.grade})"

# List of students
students = [
    Student("Alice", 22, 88.5),
    Student("Bob", 20, 91.0),
    Student("Charlie", 23, 85.0),
    Student("David", 21, 91.0)
]

print("Original list:")
print(students)
```

```python
students_by_name = sorted(students, key=lambda s: s.name)
print("Sorted by name:")
print(students_by_name)
```

```python
students_by_age = sorted(students, key=lambda s: s.age)
print("Sorted by age:")
print(students_by_age)
```

```python
students_by_grade = sorted(students, key=lambda s: s.grade, reverse=True)
print("Sorted by grade (highest first):")
print(students_by_grade)
```

```python
# Matrix with 3 columns
matrix = [
    [3, 9, 20],
    [1, 5, 10],
    [3, 7, 15],
    [2, 8, 12],
    [1, 9, 14]
]

# Sort:
# - 1st column ASCENDING
# - 2nd column DESCENDING
# - 3rd column default (ASC)
sorted_matrix = sorted(matrix, key=lambda row: (row[0], -row[1], row[2]))

print("Original Matrix:")
for row in matrix:
    print(row)

print("\nSorted Matrix (col1 asc, col2 desc):")
for row in sorted_matrix:
    print(row)
```

# 📦 Python `bisect` and `heapq` Modules

Python provides built-in modules to simplify **binary search** and **heap operations**.

---

## 1️⃣ `bisect` Module

The `bisect` module helps **insert elements into a sorted list** while maintaining the sorted order.

### 🔹 Common Functions
- `bisect.bisect_left(list, x)` → finds index to insert `x` to keep list sorted (left insertion)
- `bisect.bisect_right(list, x)` → finds index to insert `x` (right insertion)
- `bisect.insort_left(list, x)` → insert `x` at left position
- `bisect.insort_right(list, x)` → insert `x` at right position


```python
### 🔍 Example 1: Find Position to Insert

import bisect

arr = [1, 3, 4, 4, 7]
x = 4

left_index = bisect.bisect_left(arr, x)
right_index = bisect.bisect_right(arr, x)

print("Array:", arr)
print("Left insertion index of 4:", left_index)
print("Right insertion index of 4:", right_index)
```

```python
arr = [1, 3, 4, 4, 7]

# Insert 5
bisect.insort(arr, 5)
print("Array after inserting 5:", arr)

# Insert 4 on the right
bisect.insort_right(arr, 4)
print("Array after inserting 4 (right):", arr)
```

# 🔺 Python `heapq` Module

The `heapq` module provides **heap (priority queue) operations** in Python.  
By default, it implements a **min-heap**, where the smallest element is always at the root.

---

## 1️⃣ Key Functions

| Function | Description |
|----------|-------------|
| `heapq.heapify(list)` | Convert a list into a heap in-place |
| `heapq.heappush(heap, item)` | Push a new item onto the heap |
| `heapq.heappop(heap)` | Pop the smallest item from the heap |
| `heapq.heappushpop(heap, item)` | Push item, then pop the smallest element |
| `heapq.heapreplace(heap, item)` | Pop smallest, then push new item (more efficient than push + pop) |
| `heapq.nlargest(n, iterable)` | Return n largest elements |
| `heapq.nsmallest(n, iterable)` | Return n smallest elements |

---

```python
## 2️⃣ Example 1: Basic Heap Operations

import heapq

heap = [3, 1, 4, 1, 5, 9, 2]
heapq.heapify(heap)
print("Heapified list:", heap)

heapq.heappush(heap, 0)
print("After pushing 0:", heap)

smallest = heapq.heappop(heap)
print("Popped smallest element:", smallest)
print("Heap after pop:", heap)
```
