## What is `__repr__`?
`__repr__(self)` returns the **official string representation** of an object, intended for developers.
- It should be **unambiguous** and ideally could be used to **recreate** the object when possible.
- Python uses `repr(obj)` or `obj.__repr__()` in interactive consoles and sometimes in logs.
- When you `print(obj)`, Python tries `__str__` first (if defined). If not, it falls back to `__repr__`.

**Tip:** If you only implement `__repr__`, it will be used for both `repr(obj)` and `print(obj)`.



## Q1. `Point2D` — distance, midpoint, translate
**Purpose of each method:**
- `__init__(x, y)`: Construct a 2D point with coordinates `(x, y)`.
- `distance_to(other)`: Compute the Euclidean distance from this point to `other`.
- `midpoint(other)`: Return the midpoint between this point and `other` as a new `Point2D`.
- `translate(dx, dy)`: Shift (mutate) the point by the vector `(dx, dy)`.
- `__repr__()`: Developer-friendly string showing the point.


```python

class Point2D:
    def __init__(self, x=0.0, y=0.0):
        """Create a 2D point.

        Args:
            x (float): x-coordinate
            y (float): y-coordinate
        """
        self.x, self.y = float(x), float(y)

    def distance_to(self, other):
        """Return the Euclidean distance to another Point2D.

        Formula: sqrt((x1 - x2)^2 + (y1 - y2)^2)
        """
        dx, dy = self.x - other.x, self.y - other.y
        return (dx*dx + dy*dy) ** 0.5

    def midpoint(self, other):
        """Return the midpoint between self and other as a new Point2D."""
        return Point2D((self.x + other.x)/2, (self.y + other.y)/2)

    def translate(self, dx, dy):
        """Shift this point in-place by (dx, dy)."""
        self.x += dx
        self.y += dy

    def __repr__(self):
        """Unambiguous string form useful for debugging/inspection."""
        return f"Point2D({self.x:.2f}, {self.y:.2f})"
```

```python

# DEMO: Point2D
a, b = Point2D(1, 2), Point2D(4, 6)
print("a:", a, "| b:", b)
print("distance a->b:", round(a.distance_to(b), 3))
print("midpoint(a,b):", a.midpoint(b))
a.translate(-1, 3)
print("a after translate(-1,3):", a)
```


## Q2. `Rectangle` — area, perimeter, diagonal, scale
**Purpose of each method:**
- `__init__(width, height)`: Construct a rectangle with given dimensions.
- `area()`: Compute `width × height`.
- `perimeter()`: Compute `2 × (width + height)`.
- `diagonal()`: Compute the diagonal length using Pythagoras.
- `scale(factor)`: Scale the rectangle in-place by `factor`.
- `__repr__()`: Show width/height nicely for debugging.


```python

class Rectangle:
    def __init__(self, width, height):
        """Create a rectangle with the given width and height."""
        self.width, self.height = float(width), float(height)

    def area(self):
        """Return the area of the rectangle (width * height)."""
        return self.width * self.height

    def perimeter(self):
        """Return the perimeter (2 * (width + height))."""
        return 2 * (self.width + self.height)

    def diagonal(self):
        """Return the diagonal length via Pythagorean theorem."""
        return (self.width**2 + self.height**2) ** 0.5

    def scale(self, factor):
        """Scale both width and height by 'factor' in-place."""
        self.width *= factor
        self.height *= factor

    def __repr__(self):
        """Developer-friendly view of this rectangle."""
        return f"Rectangle(w={self.width:.2f}, h={self.height:.2f})"
```

```python

# DEMO: Rectangle
r = Rectangle(3, 4)
print(r)
print("area:", r.area())
print("perimeter:", r.perimeter())
print("diagonal:", round(r.diagonal(), 3))
r.scale(1.5)
print("after scale(1.5):", r, "| new area:", r.area())
```


## Q3. `RunningAverage` — mean, count, reset
**Purpose of each method:**
- `__init__()`: Initialize counters for number of values `n` and `total` sum.
- `add(x)`: Add a new data point to the stream and update `n` and `total`.
- `mean()`: Return the current average (or 0.0 if empty).
- `reset()`: Reset `n` and `total` to start fresh.
- `__repr__()`: Quick snapshot showing sample count and current mean.


```python

class RunningAverage:
    def __init__(self):
        """Create a running-average accumulator."""
        self.n = 0
        self.total = 0.0

    def add(self, x):
        """Include a new value in the running statistics."""
        self.n += 1
        self.total += float(x)

    def mean(self):
        """Return the current average; 0.0 if no samples yet."""
        return self.total / self.n if self.n else 0.0

    def reset(self):
        """Clear all accumulated state (n and total)."""
        self.n, self.total = 0, 0.0

    def __repr__(self):
        """Developer-focused snapshot of the current state."""
        return f"RunningAverage(n={self.n}, mean={self.mean():.3f})"
```

```python

# DEMO: RunningAverage
ra = RunningAverage()
for v in [10, 12, 9, 14]:
    ra.add(v)
print(ra, "| mean:", ra.mean())
ra.reset()
ra.add(100)
print("after reset:", ra, "| mean:", ra.mean())
```


## Q4. `DiceSimulator` — roll and summarize
**Purpose of each method:**
- `__init__(sides=6)`: Create a die with `sides` faces and empty history.
- `roll(times=1)`: Simulate rolling the die multiple times and store outcomes.
- `frequency()`: Return a dictionary mapping face → count from history.
- `__repr__()`: Show how many sides and how many rolls have been recorded.


```python

import random

class DiceSimulator:
    def __init__(self, sides=6):
        """Initialize a die with the given number of sides and empty roll history."""
        self.sides = int(sides)
        self.history = []

    def roll(self, times=1):
        """Roll the die 'times' times and append results to history."""
        for _ in range(times):
            self.history.append(random.randint(1, self.sides))

    def frequency(self):
        """Return a dict of face -> count across all previous rolls."""
        freq = {face: 0 for face in range(1, self.sides + 1)}
        for v in self.history:
            freq[v] += 1
        return freq

    def __repr__(self):
        """Developer view: die configuration and how many rolls recorded."""
        return f"DiceSimulator(sides={self.sides}, rolls={len(self.history)})"
```

```python

# DEMO: DiceSimulator
ds = DiceSimulator()
ds.roll(20)
print(ds)
print("frequency:", ds.frequency())
```


## Q5. `TextStats` — words, letters, top letter
**Purpose of each method:**
- `__init__(text)`: Store the original text.
- `word_count()`: Count words by splitting on whitespace and ignoring empty tokens.
- `letter_count()`: Count only alphabetic characters (A–Z/a–z).
- `most_common_letter()`: Find the most frequent letter (case-insensitive); break ties by lexicographic order; return `(letter, count)` or `(None, 0)` if none.
- `__repr__()`: Quick view showing the text length.


```python

class TextStats:
    def __init__(self, text):
        """Keep a reference to the input text as a string."""
        self.text = str(text)

    def word_count(self):
        """Return the number of whitespace-separated words (ignoring empty tokens)."""
        return len([w for w in self.text.split() if w.strip()])

    def letter_count(self):
        """Return the number of alphabetic characters in the text."""
        return sum(1 for ch in self.text if ch.isalpha())

    def most_common_letter(self):
        """Return (most_frequent_letter, count) considering only A–Z; case-insensitive.

        If multiple letters tie, return the lexicographically smallest letter.
        If no letters exist, return (None, 0).
        """
        counts = {}
        for ch in self.text.lower():
            if ch.isalpha():
                counts[ch] = counts.get(ch, 0) + 1
        if not counts:
            return None, 0
        # Choose smallest letter among those with maximum frequency
        best_letter = min((-cnt, ch) for ch, cnt in counts.items())[1]
        return best_letter, counts[best_letter]

    def __repr__(self):
        """Summarize the text object by its length in characters."""
        return f"TextStats(len={len(self.text)})"
```

```python

# DEMO: TextStats
txt = TextStats("Practice makes progress. Keep practicing!")
print("words:", txt.word_count())
print("letters:", txt.letter_count())
print("most common letter:", txt.most_common_letter())
```


## Q6. `PalindromeCleaner` — normalized palindrome check
**Purpose of each method:**
- `__init__(s)`: Store the raw string.
- `normalized()`: Return a lowercase string containing only alphanumerics from raw.
- `is_palindrome()`: Check if the normalized string reads the same forward/backward.
- `__repr__()`: Display the raw string for clarity when inspecting.


```python

class PalindromeCleaner:
    def __init__(self, s):
        """Store the original raw string."""
        self.raw = str(s)

    def normalized(self):
        """Return a lowercase string containing only [a-z0-9] from the raw input."""
        return "".join(ch.lower() for ch in self.raw if ch.isalnum())

    def is_palindrome(self):
        """Return True if the normalized string is a palindrome, else False."""
        t = self.normalized()
        return t == t[::-1]

    def __repr__(self):
        """Developer view that shows the original input string."""
        return f"PalindromeCleaner({self.raw!r})"
```

```python

# DEMO: PalindromeCleaner
p = PalindromeCleaner("A man, a plan, a canal: Panama!")
print("normalized:", p.normalized())
print("is_palindrome:", p.is_palindrome())
```


## Q7. `ShoppingCart` — line items, subtotal, tax, total
**Purpose of each method:**
- `__init__()`: Initialize empty list of items; each item is `(name, price_per_unit, quantity)`.
- `add(name, price, qty=1)`: Add a new line item to the cart.
- `subtotal()`: Sum of `price × quantity` across items.
- `total(tax_rate_percent=0.0)`: Apply tax (percentage) to subtotal and return final total.
- `receipt()`: Generate a formatted, human-readable receipt string.
- `__repr__()`: Quick snapshot of how many items are in the cart.


```python

class ShoppingCart:
    def __init__(self):
        """Initialize an empty shopping cart."""
        self.items = []  # each item is (name, price_per_unit, quantity)

    def add(self, name, price, qty=1):
        """Add an item to the cart with given name, unit price, and quantity."""
        self.items.append((str(name), float(price), int(qty)))

    def subtotal(self):
        """Compute the sum of (price * quantity) for all items in the cart."""
        return sum(price * qty for _, price, qty in self.items)

    def total(self, tax_rate_percent=0.0):
        """Return subtotal with tax applied (percentage).

        Args:
            tax_rate_percent (float): e.g., 5 means 5% tax
        """
        sub = self.subtotal()
        return sub * (1 + tax_rate_percent / 100.0)

    def receipt(self):
        """Return a multi-line string itemizing the bill and showing totals."""
        lines = [
            f"{name} x{qty} @ {price:.2f} = {price*qty:.2f}"
            for name, price, qty in self.items
        ]
        lines.append(f"Subtotal: {self.subtotal():.2f}")
        lines.append(f"Total (5% tax): {self.total(5):.2f}")
        return "\n".join(lines)

    def __repr__(self):
        """Developer snapshot of how many items the cart contains."""
        return f"ShoppingCart({len(self.items)} items)"
```

```python

# DEMO: ShoppingCart
cart = ShoppingCart()
cart.add("Notebook", 49.99, 2)
cart.add("Pen", 15.5, 3)
print(cart)
print(cart.receipt())
```


## Q8. `Temperature` — Celsius/Fahrenheit conversion
**Purpose of each method:**
- `__init__(celsius)`: Store the temperature in Celsius.
- `to_fahrenheit()`: Convert and return Fahrenheit value.
- `add_offset(dc)`: Increase/decrease Celsius value by `dc`.
- `__repr__()`: Show a nicely formatted Celsius and Fahrenheit string.


```python

class Temperature:
    def __init__(self, celsius):
        """Create a Temperature object storing Celsius internally."""
        self.c = float(celsius)

    def to_fahrenheit(self):
        """Convert the stored Celsius temperature to Fahrenheit."""
        return self.c * 9/5 + 32

    def add_offset(self, dc):
        """Adjust the Celsius temperature by dc (delta Celsius)."""
        self.c += float(dc)

    def __repr__(self):
        """Human-friendly display of both Celsius and Fahrenheit."""
        return f"{self.c:.1f}°C ({self.to_fahrenheit():.1f}°F)"
```

```python

# DEMO: Temperature
t = Temperature(24)
print(t)
t.add_offset(3.5)
print("after +3.5°C:", t, "| in °F:", t.to_fahrenheit())
```


## Q9. `BudgetTracker` — add expenses, totals by category
**Purpose of each method:**
- `__init__()`: Initialize empty entries list; each entry is `(category, amount)`.
- `add(category, amount)`: Add a new expense to the tracker.
- `total()`: Total spend across all categories.
- `by_category()`: Aggregate spend per category and sort descending by spend (then name).
- `top_category()`: Return the highest-spend category tuple or `(None, 0.0)` if empty.
- `__repr__()`: Snapshot of how many spending entries are recorded.


```python

class BudgetTracker:
    def __init__(self):
        """Create a tracker with no entries."""
        self.entries = []  # (category, amount)

    def add(self, category, amount):
        """Record an expense with a category and a numeric amount."""
        self.entries.append((str(category), float(amount)))

    def total(self):
        """Return the total amount spent across all entries."""
        return sum(a for _, a in self.entries)

    def by_category(self):
        """Return a sorted list of (category, total_spend) by descending spend, then name."""
        agg = {}
        for cat, amt in self.entries:
            agg[cat] = agg.get(cat, 0.0) + amt
        # sort primarily by -spend, secondarily by category name
        return sorted(agg.items(), key=lambda kv: (-kv[1], kv[0]))

    def top_category(self):
        """Return the category with the highest total spend (or (None, 0.0) if none)."""
        bc = self.by_category()
        return bc[0] if bc else (None, 0.0)

    def __repr__(self):
        """Developer snapshot of how many entries exist."""
        return f"BudgetTracker({len(self.entries)} entries)"
```

```python

# DEMO: BudgetTracker
bt = BudgetTracker()
bt.add("Food", 250); bt.add("Transport", 80); bt.add("Food", 120); bt.add("Rent", 900)
print(bt)
print("Total:", bt.total())
print("By category:", bt.by_category())
print("Top category:", bt.top_category())
```


## Q10. `TimeSpan` — add durations, pretty HH:MM:SS
**Purpose of each method:**
- `__init__(seconds=0)`: Store total duration in seconds.
- `add(hours=0, minutes=0, seconds=0)`: Add a duration to the current total.
- `total_seconds()`: Return the integer number of seconds stored.
- `pretty()`: Format the stored duration as `HH:MM:SS` (zero-padded).
- `__repr__()`: Developer-friendly display that reuses `pretty()`.


```python

class TimeSpan:
    def __init__(self, seconds=0):
        """Create a duration container, stored internally in total seconds."""
        self._s = int(seconds)

    def add(self, hours=0, minutes=0, seconds=0):
        """Add a duration to the total using hours/minutes/seconds."""
        self._s += int(hours) * 3600 + int(minutes) * 60 + int(seconds)

    def total_seconds(self):
        """Return the total stored seconds as an integer."""
        return self._s

    def pretty(self):
        """Return a zero-padded string in HH:MM:SS format."""
        s = self._s
        h, s = divmod(s, 3600)
        m, s = divmod(s, 60)
        return f"{h:02d}:{m:02d}:{s:02d}"

    def __repr__(self):
        """Developer view of the duration using the pretty() formatter."""
        return f"TimeSpan({self.pretty()})"
```

```python

# DEMO: TimeSpan
ts = TimeSpan()
ts.add(minutes=45)
ts.add(hours=1, minutes=30)
ts.add(seconds=40)
print(ts, "| total seconds:", ts.total_seconds())
print("pretty:", ts.pretty())
```
