# Tutorial: List Comprehensions, Dictionaries

### Q1. Matrix Transpose

```python
matrix = [[1, 2, 3], [4, 5, 6]]
transpose = [[row[i] for row in matrix] for i in range(len(matrix[0]))]
print(matrix)
print(transpose)
```

### Q2. Dictionary of Word Lengths

```python
words = ['apple','banana','cherry']
word_lengths = {w: len(w) for w in words}
print(word_lengths) 
```

### Q3. Filter Even Squares

```python
nums = list(range(1,21))
even_squares = [n**2 for n in nums if n%2==0]
print(nums)
print(even_squares)
```

### Q4. Character Frequency

```python
s = 'practice good habits daily'
counts = {c: s.count(c) for c in set(s) if c.isalpha()}
print(counts)
```

### Q5. Swap Keys & Values

```python
d = {'a':1,'b':2,'c':3}
inverted = {v:k for k,v in d.items()}
print(inverted)
```

### Q6. Combine Lists into Dict

```python
names = ['A','B','C']
marks = [88,79,95]
record = {n:m for n,m in zip(names,marks)}
print(record)
```

### Q7. Filter Dict by Condition

```python
scores = {'A':80,'B':67,'C':90}
passed = {k:v for k,v in scores.items() if v>=70}
print(passed)
```

### Q8. Flatten Nested Lists

```python
nested = [[1,2],[3,4],[5,6]]
flat = [x for sub in nested for x in sub]
print(flat)
```

### Q9. Dict from Pairs with Conditions

```python
pairs = [('a',1),('b',2),('c',3)]
dict1 = {k:v for k,v in pairs if v%2!=0}
print(dict1)
```

### Q10. Nested Comprehension for Table

```python
table = [[i*j for j in range(1,6)] for i in range(1,6)]
print(table)
for row in table: 
    print(row)
```
