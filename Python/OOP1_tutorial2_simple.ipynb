## Q1. Distance2D — Compute distance between two points

```python

class Distance2D:
    def __init__(self, x1, y1, x2, y2):
        self.x1, self.y1, self.x2, self.y2 = float(x1), float(y1), float(x2), float(y2)
    def compute(self):
        dx, dy = self.x2 - self.x1, self.y2 - self.y1
        return (dx**2 + dy**2) ** 0.5

# demo
p = Distance2D(1, 2, 4, 6)
print("Distance =", round(p.compute(), 3))
```

## Q2. CircleArea — Find area and circumference

```python

import math

class CircleArea:
    def __init__(self, radius):
        self.radius = float(radius)
    def area(self):
        return math.pi * self.radius**2
    def circumference(self):
        return 2 * math.pi * self.radius

# demo
c = CircleArea(5)
print("Area:", round(c.area(), 2), "| Circumference:", round(c.circumference(), 2))
```

## Q3. WordAnalyzer — Count vowels and consonants

```python

class WordAnalyzer:
    def __init__(self, text):
        self.text = text.lower()
    def vowels(self):
        return sum(1 for ch in self.text if ch in "aeiou")
    def consonants(self):
        return sum(1 for ch in self.text if ch.isalpha() and ch not in "aeiou")

wa = WordAnalyzer("Education")
print("Vowels:", wa.vowels(), "| Consonants:", wa.consonants())
```

## Q4. SimpleCalculator — Perform add, subtract, multiply, divide

```python

class SimpleCalculator:
    def __init__(self, a, b):
        self.a, self.b = float(a), float(b)
    def add(self): return self.a + self.b
    def subtract(self): return self.a - self.b
    def multiply(self): return self.a * self.b
    def divide(self): return self.a / self.b if self.b != 0 else None

# demo
calc = SimpleCalculator(12, 4)
print("Add:", calc.add(), "| Subtract:", calc.subtract(), "| Multiply:", calc.multiply(), "| Divide:", calc.divide())
```

## Q5. TemperatureConverter — Convert between Celsius and Fahrenheit

```python

class TemperatureConverter:
    def __init__(self, celsius):
        self.c = float(celsius)
    def to_fahrenheit(self):
        return (self.c * 9/5) + 32
    def to_kelvin(self):
        return self.c + 273.15

# demo
t = TemperatureConverter(30)
print("Celsius:", t.c, "| Fahrenheit:", t.to_fahrenheit(), "| Kelvin:", t.to_kelvin())
```
